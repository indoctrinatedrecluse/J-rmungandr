# 🪐 Jörmungandr Reactive DAG & AST Analysis Demo

This sample notebook is designed to test Jörmungandr's **Reactive Execution DAG**,
**Variable Dependency Tracking**, and **Out-of-Order Execution Detection**.

### How to Test:
1. Run Cells 1, 2, and 3 in sequential order.
2. Observe the **Reactive DAG Canvas** in the tool window update with node dependencies.
3. Go back and modify `base_scalar` in Cell 1 without running Cell 2.
4. Notice the **⚠️ Stale** badge appears on downstream dependent cells.
5. Click **⚡ Topological Run** to re-execute in clean dependency order.

```python
# Cell 1: Defines base inputs and parameters
base_scalar = 42
scaling_factor = 1.5
label = 'Sensor-Alpha'
print(f'Base configuration initialized: scalar={base_scalar}, factor={scaling_factor}')
```

```python
# Cell 2: Consumes base_scalar & scaling_factor, defines transformed_vector
transformed_vector = [int((x + base_scalar) * scaling_factor) for x in range(10)]
print(f'Transformed vector of {len(transformed_vector)} points generated.')
```

```python
# Cell 3: Consumes label and transformed_vector, computes metrics
summary_metrics = {
    'label': label,
    'count': len(transformed_vector),
    'mean': sum(transformed_vector) / len(transformed_vector),
    'sum': sum(transformed_vector)
}

print(f"Summary for {summary_metrics['label']}:")
print(f"  Count: {summary_metrics['count']}")
print(f"  Mean:  {summary_metrics['mean']}")
print(f"  Sum:   {summary_metrics['sum']}")
```

### Independent Branch
Cell 4 below does NOT depend on Cell 1, 2, or 3. In the reactive DAG, it forms an independent root node.

```python
# Cell 4: Independent branch
import random
random.seed(1337)
random_stream = [random.randint(10, 100) for _ in range(8)]
print(f'Independent random stream: {random_stream}')
```

### [DAG] Programmatic Reactive DAG Activation
Run the cell below to programmatically open and focus the interactive Jormungandr Reactive DAG canvas.

```python
# Ensure samples directory is in sys.path
import sys, os
for _p in [os.path.abspath('..'), os.path.abspath('.'), os.path.abspath('../..')]:
    if _p not in sys.path:
        sys.path.insert(0, _p)
try:
    import jormungandr as jm
except ImportError:
    import json
    class _FallbackJm:
        @staticmethod
        def _trigger(action, **kwargs):
            payload = {'action': action, **kwargs}
            print('__JG_IDE_ACTION__' + json.dumps(payload) + '__JG_IDE_ACTION_END__', flush=True)
            print(f"[Jormungandr Studio Trigger] {kwargs.get('title', action)} -> Studio activated.")
        @classmethod
        def show_dataframe(cls, *a, **kw): cls._trigger('show_dataframe', **kw)
        @classmethod
        def show_lakehouse(cls, *a, **kw): cls._trigger('show_lakehouse', **kw)
        @classmethod
        def show_pipeline_lineage(cls, *a, **kw): cls._trigger('show_pipeline_lineage', **kw)
        @classmethod
        def show_database_studio(cls, *a, **kw): cls._trigger('show_database_studio', **kw)
        @classmethod
        def show_model_inspector(cls, *a, **kw): cls._trigger('show_model_inspector', **kw)
        @classmethod
        def show_gpu_monitor(cls, *a, **kw): cls._trigger('show_gpu_monitor', **kw)
        @classmethod
        def show_prompt_studio(cls, *a, **kw): cls._trigger('show_prompt_studio', **kw)
        @classmethod
        def show_r_console(cls, *a, **kw): cls._trigger('show_r_console', **kw)
        @classmethod
        def show_plots(cls, *a, **kw): cls._trigger('show_plots', **kw)
        @classmethod
        def show_dag(cls, *a, **kw): cls._trigger('show_dag', **kw)
    jm = _FallbackJm()
    sys.modules['jormungandr'] = jm
    sys.modules['jm'] = jm

# Trigger Jormungandr Reactive DAG tool window
jm.show_dag()
print('[Jormungandr] Reactive DAG tool window successfully activated!')
```
