# 🪐 Jörmungandr Reactive DAG & AST Analysis Demo

This sample notebook is designed to test Jörmungandr's **Reactive Execution DAG**,
**Variable Dependency Tracking**, and **Out-of-Order Execution Detection**.

### How to Test:
1. Run Cells 1, 2, and 3 in sequential order.
2. Observe the **Reactive DAG Canvas** in the tool window update with node dependencies.
3. Go back and modify `base_scalar` in Cell 1 without running Cell 2.
4. Notice the **⚠️ Stale** badge appears on downstream dependent cells.
5. Click **⚡ Topological Run** to re-execute in clean dependency order.

```python
# Cell 1: Defines base inputs and parameters
base_scalar = 42
scaling_factor = 1.5
label = 'Sensor-Alpha'
print(f'Base configuration initialized: scalar={base_scalar}, factor={scaling_factor}')
```

```python
# Cell 2: Consumes base_scalar & scaling_factor, defines transformed_vector
transformed_vector = [int((x + base_scalar) * scaling_factor) for x in range(10)]
print(f'Transformed vector of {len(transformed_vector)} points generated.')
```

```python
# Cell 3: Consumes label and transformed_vector, computes metrics
summary_metrics = {
    'label': label,
    'count': len(transformed_vector),
    'mean': sum(transformed_vector) / len(transformed_vector),
    'sum': sum(transformed_vector)
}

print(f"Summary for {summary_metrics['label']}:")
print(f"  Count: {summary_metrics['count']}")
print(f"  Mean:  {summary_metrics['mean']}")
print(f"  Sum:   {summary_metrics['sum']}")
```
